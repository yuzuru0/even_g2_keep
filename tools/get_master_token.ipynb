# 🔑 Google Keep Master Token Generator for Even Realities G2

This Google Colab notebook lets you generate your `GOOGLE_MASTER_TOKEN` directly in the browser — **no local Python installation or terminal required**.

---

### 📋 Step 1: Get `oauth_token` from Google Chrome
1. Open Google Chrome (or Edge) on desktop.
2. Press `F12` (or right-click -> Inspect) to open **Developer Tools**.
3. Go to **Application** > **Cookies** > `https://accounts.google.com`.
4. Open [https://accounts.google.com/EmbeddedSetup](https://accounts.google.com/EmbeddedSetup) in that tab and sign in.
   *(Note: The page hanging or showing a blank/white screen after login is **expected behavior**. Do not wait for it to load.)*
5. In the cookies list, double-click and copy the value of `oauth_token` (starts with `oauth2_4/...`).

```python
import getpass
import gpsoauth
import gkeepapi

print("=" * 60)
print(" Google Keep Master Token Generator (Even G2)")
print("=" * 60)

email = input("Google Email: ").strip()
oauth_token = getpass.getpass("Paste oauth_token (input is hidden): ").strip()

if not email or not oauth_token:
    print("Error: Email and oauth_token are required.")
else:
    print("\n[1/2] Exchanging oauth_token with Google servers...")
    android_id = "0123456789abcdef"
    res = gpsoauth.exchange_token(email, oauth_token, android_id)

    if "Token" in res:
        master_token = res["Token"]
        print("[2/2] Testing connection to Google Keep...")
        try:
            keep = gkeepapi.Keep()
            keep.resume(email, master_token)
            keep.sync()
            count = len(list(keep.all()))
            print("\n" + "★" * 60)
            print(f"🎉 SUCCESS! Connected to Google Keep. Found {count} notes.")
            print("★" * 60)
            print("\nCopy and paste this into your Render Environment Variables:")
            print("-" * 60)
            print(f"GOOGLE_MASTER_TOKEN={master_token}")
            print("-" * 60)
        except Exception as e:
            print(f"Keep verification failed: {e}")
            print(f"Generated token: {master_token}")
    else:
        print(f"\n❌ Token exchange failed: {res.get('Error', res)}")
        print("Please verify that the oauth_token was copied correctly and has not expired.")
```
